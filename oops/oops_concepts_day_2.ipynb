```python
import math
class point:
    def __init__(self,x,y):
        self.x_cod=x
        self.y_cod=y
    def __str__(self):
        return '<{},{}>'.format(self.x_cod,self.y_cod)
    def euclidian_distance(self,other):
        print(math.sqrt((self.x_cod-other.x_cod)**2+(self.y_cod-other.y_cod)**2))
        
obj1=point(0,0)
print(obj1)
obj2=point(10,10)
print(obj2)
obj1.euclidian_distance(obj2)
```
