```python
class fraction:
    def __init__(self,num,den):
        self.num=num
        self.den=den
    def __str__(self):
        return "{}/{}".format(self.num,self.den)
    def __add__(self,other):
        return "{}/{}".format(other.num*self.den+other.den*self.num,self.den*other.den)
    def __sub__(self,other):
        return '{}/{}'.format(other.den*self.num-other.num*self.den,self.den*other.den)
    def __mul__(self,other):
        return '{}/{}'.format(other.num*self.num,other.den*self.den)
    def __truediv__(self,other):
        return '{}/{}'.format(self.num*other.den,self.den*other.num)

        
obj1=fraction(1,4)
obj2=fraction(3,4)
print(obj1+obj2)
print(obj1-obj2)
print(obj1*obj2)
print(obj1/obj2)        
    
```
